```python
#Change timeformat to remove microsecconds
df['drop_time'] = pd.to_datetime(df['drop_time'].dt.strftime('%Y-%m-%d %H:%M:%S'))
```

```python
#Add a new column to find out the total time taken to complete a ride
df['total ride time']=df['drop_time']-df['pickup_time']
#convert timedelta format string by removing days
df['total ride time']=df['total ride time'].astype(str).str.replace('0 days','',regex=False)
```
